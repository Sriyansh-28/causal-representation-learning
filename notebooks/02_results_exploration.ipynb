## Per-replicate PEHE

IHDP realizations differ enormously in outcome scale, so the mean is
dominated by a few replicates. Plot on a log scale to see the typical case.

```python
fig, ax = plt.subplots(figsize=(6, 3.5))
for model, grp in df.groupby("model"):
    ax.plot(grp["seed"], grp["pehe"], marker="o", label=model, alpha=0.8)
ax.set_yscale("log"); ax.set_xlabel("realization (seed)"); ax.set_ylabel("PEHE")
ax.legend(fontsize=8); ax.set_title("Per-realization PEHE, IHDP")
plt.show()
```
